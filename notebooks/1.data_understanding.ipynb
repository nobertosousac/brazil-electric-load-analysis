## Entendimento dos Dados

Este notebook será utilizado para analisar um recorte de dados referente ao ano de 2025. Antes de fazer toda a carga histórica, será importado apenas dados de 2025 para analisar a estrutura do dataset, quais são as colunas disponíveis, o que cada uma delas está representando e se realmente está de acordo com o que precisamos para o desenvolvimento deste projeto.

### 1. Importação das Bibliotecas

```python
from pathlib import Path
import pandas as pd

DATA_PATH = Path('../data/raw')
```

### 2. Carregando e Compreendendo os Dados

```python
electric_load_2025 = pd.read_parquet(DATA_PATH / 'curva_carga_2025.parquet')
electric_load_2025.head()
```

#### 2.1. Dicionário do Dataset

Cada linha está representando uma carga elétrica que foi registrada em um subsistema durante uma determinada hora. Então, por exemplo, no dia 1 de Janeiro de 2025, à meia-noite, o subsistema NORDESTE registrou uma carga de 7097.75 MWmed, isso quer dizer que, naquele horário, o sistema precisou atender uma demanda média de aproximadamente 7 mil megawatts para aquela determinada região.

Abaixo está cada coluna existente e o que cada uma está trazendo de informação:

| Coluna | Descrição |
|---|---|
| id_subsistema | Sigla do subsistema elétrico |
| nom_subsistema | Nome do subsistema |
| din_instante | Data e hora de referência |
| val_cargaenergiahomwmed | Carga elétrica média no período, medida em MWmed |

#### 2.2. Verificando a Dimensão

```python
print(f'O dataset possui um total de {electric_load_2025.shape[0]} linhas e {electric_load_2025.shape[1]} colunas.')
```

#### 2.3. Informações Gerais

```python
electric_load_2025.info(memory_usage='deep')
```

#### 2.4. Confirmação de Anos e Horários

Nesta etapa, será verificado o ano inicial (menor ano) e o ano final (maior ano), para entender se o recorte de dados que fizemos está representando corretamento o intervalo que selecionamos. Além disso, também será observado a quantidade de horários distintos que temos.

```python
print(f"Data inicial: {electric_load_2025['din_instante'].min()}")
print(f"Data final: {electric_load_2025['din_instante'].max()}")
print(f"Horários distintos: {electric_load_2025['din_instante'].nunique():,}")
```

#### 2.5. Verificando Valores Únicos dos Subsistemas

Agora, iremos observar quais são os subsistemas existentes em nosso conjunto de dados, eliminando qualquer tipo de duplicidade e mantendo apenas os registros que aparecem pelo menos uma vez nesse recorte de dados.

```python
subsystems = electric_load_2025[['id_subsistema', 'nom_subsistema']].drop_duplicates().sort_values('id_subsistema')

subsystems
```

#### 2.6. Frequência Absoluta dos Subsistemas

Abaixo, teremos a frequência que cada subsistema aparece no conjunto de dados.

```python
electric_load_2025['nom_subsistema'].value_counts()
```
